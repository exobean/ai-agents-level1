## Roles

```
from shopassist.client import post_json
from shopassist.mockserver import MockProvider

question = [{"role": "user", "content": "Where is order #1042?"}]
systems = ["You are ShopAssist.",
           "You are ShopAssist. Be brief.",
           "You are ShopAssist. Sign every reply."]

with MockProvider() as server:
    for system in systems:
        body = {"model": "mock", "max_tokens": 200,
                "system": system, "messages": question}
        reply = post_json(server.url, body, "test-key")
        print("system:", system)
        print("  reply:", reply["content"][0]["text"])
```
