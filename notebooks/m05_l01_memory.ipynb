## Memory

```
import textwrap

from shopassist.app_v2 import new_session


def show(label, text):
    print(textwrap.fill(text, 68, initial_indent=label,
                        subsequent_indent=" " * len(label)))


chat, _ = new_session()
show("A1: ", chat.ask("Where is my order #1042?"))
show("A2: ", chat.ask("Can I still change its address?"))
print("messages kept so far:", len(chat.messages))

# The same second question in a brand-new conversation:
fresh, _ = new_session()
show("new chat: ", fresh.ask("Can I still change its address?"))
```
