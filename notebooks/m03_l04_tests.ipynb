## Golden

```
from shopassist.prompting import CASES, classify, load_prompt


def golden_test(version: int) -> None:
    prompt = load_prompt("intent", version)
    bad = [(t, w, classify(prompt, t)) for t, w in CASES
           if classify(prompt, t) != w]
    status = "PASS" if not bad else f"FAIL ({len(bad)} case)"
    print(f"intent v{version}: {status}")
    for text, want, got in bad:
        print(f"   {text!r}")
        print(f"   expected {want}, got {got}")


for version in (3, 4):
    golden_test(version)
```
