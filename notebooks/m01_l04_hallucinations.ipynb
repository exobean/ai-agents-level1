## Hallucinate

```
from shopassist.agent import run_agent
from shopassist.models import TeachingModel
from shopassist.tools import GET_ORDER, ORDERS
from shopassist.toylm import BigramModel

print("is order 1045 in the database?", "1045" in ORDERS)

# 1. Pattern completion: the toy model just continues the text.
guess = BigramModel().next_probs("is")
best = max(guess, key=guess.get)
print(f"toy model: 'order 1045 is {best}'", f"(p = {guess[best]:.2f})")

# 2. Grounded: the agent looks the order up with a tool first.
answer, _ = run_agent(TeachingModel(), [GET_ORDER], "Be accurate.",
                      "Where is order #1045?")
print("grounded agent:", answer.split(". ")[0] + ".")
```
