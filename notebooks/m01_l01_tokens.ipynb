## Tokens

```
from shopassist.toylm import CORPUS, tokenize, train_bpe

for merges in (0, 10, 30):
    rules = train_bpe(CORPUS, merges)
    pieces = tokenize("order 1042 is packed", rules)
    shown = "|".join(pieces)
    print(f"{merges:>2} merges, {len(pieces):>2} tokens: {shown}")
```
