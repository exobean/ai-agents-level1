## Function call

```
import json

from shopassist.messages import Message
from shopassist.models import TeachingModel
from shopassist.providers import to_request
from shopassist.tools import GET_ORDER

# 1. What the model is told: the tool definitions.
question = [Message("user", "Is order #1043 on its way?")]
body = to_request("You are ShopAssist.", question, [GET_ORDER])
tool = body["tools"][0]
print("name:       ", tool["name"])
print("description:", tool["description"])
schema = json.dumps(tool["input_schema"], indent=1)
print("input_schema:", schema)

# 2. What the model sends back: a request, not a result.
model = TeachingModel()
reply = model.reply("You are ShopAssist.", question, [GET_ORDER])
call = reply.tool_calls[0]
print("model asks for:", call.name, call.args)
```
