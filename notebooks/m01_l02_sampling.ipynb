## Sampling

```
from shopassist.toylm import BigramModel

model = BigramModel()
for t in (0.2, 1.0, 2.0):
    probs = model.next_probs("is", temperature=t)
    top3 = list(probs.items())[:3]
    shown = ", ".join(f"{w} {p:.2f}" for w, p in top3)
    print(f"temperature {t}: {shown}")

print()
for t in (0, 1.0, 2.0):
    for seed in (1, 2):
        words = model.generate("order", 4, temperature=t, seed=seed)
        print(f"T={t} seed={seed}:", " ".join(words))
```
