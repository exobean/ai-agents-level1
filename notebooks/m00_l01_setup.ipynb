## Grade

```
import re
import shutil
import subprocess
import sys
import tempfile
from pathlib import Path

import shopassist

COMPANION = Path(shopassist.__file__).parents[1]


def grade(folder: Path) -> str:
    cmd = [sys.executable, "-m", "pytest", "-rN",
           "-p", "no:cacheprovider", "exercises/m04_l02"]
    out = subprocess.run(
        cmd,
        cwd=folder, capture_output=True, text=True)
    last = out.stdout.strip().splitlines()[-1]
    return re.sub(r" in [0-9.]+s", "", last)   # drop the timing


print("starter, before you write anything:", grade(COMPANION))

with tempfile.TemporaryDirectory() as tmp:
    copy = Path(tmp) / "companion"
    shutil.copytree(COMPANION, copy)
    answer = copy / "solutions" / "m04_l02" / "my_solutions.py"
    shutil.copy(answer, copy / "exercises" / "m04_l02")
    print("after a correct answer:            ", grade(copy))
```
