## Cost

```
import json

from shopassist.agent import run_agent
from shopassist.costs import cost_rupees, estimate_tokens
from shopassist.models import TeachingModel
from shopassist.providers import to_request
from shopassist.tools import GET_ORDER

RATE = 250   # rupees per million tokens: a made-up example rate

system = "You are ShopAssist, a polite support agent."
model = TeachingModel()
_, msgs = run_agent(model, [GET_ORDER], system, "Order #1042?")

# The agent called the model twice: with 1 message, then with 3.
sent = [to_request(system, msgs[:n], [GET_ORDER]) for n in (1, 3)]
tokens = sum(estimate_tokens(json.dumps(body)) for body in sent)

one = cost_rupees(tokens, RATE)
print("tokens sent, about:      ", tokens)
print("cost per question, Rs:   ", round(one, 4))
print("cost per 1000 questions: ", round(one * 1000, 2))
```
