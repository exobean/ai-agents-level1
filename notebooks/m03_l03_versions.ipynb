## Versions

```
from shopassist.prompting import CASES, accuracy, load_prompt

for version in (1, 2, 3, 4):
    prompt = load_prompt("intent", version)
    header = prompt.splitlines()[0]            # "# intent vN - why"
    score = accuracy(prompt, CASES)
    print(f"{score:>4.0%}  {header[2:]}")
```
