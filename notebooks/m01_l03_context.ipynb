## Context

```
from shopassist.context import fit_to_window, message_tokens
from shopassist.sample_chat import LONG_CHAT

total = sum(message_tokens(m) for m in LONG_CHAT)
print("whole chat:", len(LONG_CHAT), "messages,", total, "tokens")

for keep_first in (False, True):
    kept = fit_to_window(LONG_CHAT, 70, keep_first=keep_first)
    knows = any("1042" in m.text for m in kept)
    print(f"keep_first={keep_first!s:<5} kept {len(kept)} messages;",
          "order number still visible:", knows)
```
