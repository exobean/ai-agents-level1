## Tool choice

```
from shopassist.tool_questions import GOOD, QUESTIONS, VAGUE
from shopassist.toolkit import choose_tool

for label, tools in (("vague", VAGUE), ("clear", GOOD)):
    right = sum(choose_tool(q, tools) == want for q, want in QUESTIONS)
    print(f"{label} descriptions: {right} of {len(QUESTIONS)} right")
    for t in tools:
        d = t.description
        print(f"   {t.name}:", d if len(d) <= 40 else d[:40] + "...")
```
