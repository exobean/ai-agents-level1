## Eval

```
from shopassist.app_v2 import new_session
from shopassist.evals import EVAL_SET, run_eval, wilson
from shopassist.messages import Message
from shopassist.models import ScriptedModel

unsure = Message("assistant", "Sorry, I am not sure.")
agents = {
    "ShopAssist v2": lambda: new_session()[0],
    "weak model": lambda: new_session(ScriptedModel([unsure]))[0],
}
for name, make in agents.items():
    passed = sum(ok for _, ok in run_eval(make))
    lo, hi = wilson(passed, len(EVAL_SET))
    print(f"{name:<14} {passed:>2}/{len(EVAL_SET)}",
          f"  95% interval {lo:.0%} to {hi:.0%}")
```
