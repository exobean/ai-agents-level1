## Faq

```
import textwrap

from shopassist.app_v2 import new_session
from shopassist.faq import search_faq

for q in ("How long does delivery take?", "Do you sell bicycles?"):
    hit = search_faq(q)
    print(q, "->", hit.get("topic", "no match"))

chat, _ = new_session()
for q in ("Do you accept UPI?", "Do you sell bicycles?"):
    print(textwrap.fill(chat.ask(q), 68, initial_indent="A: ",
                        subsequent_indent="   "))
```
