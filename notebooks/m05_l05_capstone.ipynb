## Capstone

```
import textwrap

from shopassist.app_v2 import new_session

chat, tracer = new_session()
for question in [
    "Hi, where is my order #1042?",
    "Can I still change its address?",
    "Please change the address of order #1042 to 12 MG Road.",
    "Do you accept UPI?",
    "What is your return policy?",
    "This is unacceptable, I want a manager!",
]:
    print("Q:", question)
    print(textwrap.fill(chat.ask(question), 68, initial_indent="A: ",
                        subsequent_indent="   "))
print("trace events recorded:", len(tracer.events))
```
