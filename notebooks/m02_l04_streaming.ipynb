## Stream

```
from shopassist.client import stream_text
from shopassist.mockserver import MockProvider

body = {"model": "mock", "max_tokens": 200, "system": "Be helpful.",
        "messages": [{"role": "user",
                      "content": "Where is order #1042?"}]}

pieces = []
with MockProvider() as server:
    for piece in stream_text(server.url, body, "test-key"):
        pieces.append(piece)   # a real app would show it at once

print("pieces received:", len(pieces))
print("first piece:    ", repr(pieces[0]))
print("as they arrive: ", "|".join(pieces))
print("joined:         ", "".join(pieces))
```
